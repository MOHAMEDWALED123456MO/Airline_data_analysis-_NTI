# Airlines Flights Data Analysis



```python
import pandas as pd
import matplotlib.pyplot as plt

%matplotlib inline
```

## 1. Read Data

```python

import pandas as pd

dataset = pd.read_csv(r"D:\التنزيلات\archive (10)\airlines_flights_data.csv")

dataset = dataset.drop("index", axis=1, errors="ignore")

dataset.head()

```

```python
dataset.info()
```

```python
dataset.isnull().sum()
```

## 2. Count Airlines

```python
dataset["airline"].value_counts()
```

## 3. Average Price by Airline

```python
dataset.groupby("airline")["price"].mean().sort_values()
```

```python
dataset.groupby("airline")["price"].mean().sort_values().plot(
    kind="bar",
    title="Average Price by Airline"
)
plt.show()
```

## 4. Average Price by Class

```python
dataset.groupby("class")["price"].mean()
```

## 5. Average Price by Stops

```python
dataset.groupby("stops")["price"].mean()
```

```python
dataset.groupby("stops")["price"].mean().plot(
    kind="bar",
    title="Average Price by Stops"
)
plt.show()
```
